# Impedance check

This notebook checks the impedance of the electrodes connected to the Portiloop.

Each electrode is displayed as a colored tile:
- **green**: impedance below the lower threshold (good contact),
- **yellow**: impedance between the two thresholds (acceptable, can be improved),
- **red**: impedance above the upper threshold, or **OFF** when the electrode is disconnected.

The thresholds can be changed with the `Yellow (kΩ)` slider at any time, the colors are updated immediately.

**How it works:** the ADS1299 injects a tiny AC current (6 nA or 24 nA, square wave at 31.25 Hz) in each electrode (`ch1`, `ch2`, ...) through its lead-off current sources.
The current returns through the ground (bias) electrode, and the impedance is the amplitude of the resulting 31.25 Hz voltage divided by the current.
The 5 kΩ resistor in series with each electrode on the Portiloop PCB is subtracted.

**Reference electrode (`REF`):** its impedance cannot be measured on Portiloop, because it only reaches the ADS1299 through SRB1, which has no lead-off current source.
Only its connection is checked: when the reference is unplugged, the connected electrodes saturate. `REF` then shows **OFF** (and the electrodes show **?**).
`REF` shows **connected** as long as at least one electrode is measured normally, and **?** when all electrodes are off.

**Notes:**
- The ground (bias) electrode must be connected: the measurement current returns through it. Its own impedance is not measured.
- If every electrode shows **OFF**, check the ground electrode first.
- An electrode showing **?** with `REF` connected gave an implausible measurement (e.g., an electrode shorted to another one): check it and wait for the next measurement.
- Unselect the channels that have no electrode, they would show as **OFF**.
- Do not run the impedance check while a recording is running (e.g., in the Jupyter UI or Simple GUI): both would use the ADS1299 at the same time.
- When this notebook is not running on a Portiloop, a simulated ADS1299 is used so the interface can be tried out.

```python
# Run this cell to launch the impedance check.

from portiloop.src.jupyter_gui.impedance_gui import ImpedanceCheckUI

# Default thresholds (kOhm) between green/yellow and yellow/red:
THRESHOLDS_KOHM = (10, 25)

# Resistance in series with each electrode on the PCB (Ohm), subtracted from the measurements
# (R12 to R19 on the Portiloop v2.3 PCB):
SERIES_RESISTANCE = 5000.0

ui = ImpedanceCheckUI(thresholds_kohm=THRESHOLDS_KOHM,
                      series_resistance=SERIES_RESISTANCE)
```
